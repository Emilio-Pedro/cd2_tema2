```python
import pandas as pd

df = pd.read_csv(
    "../ubernymaio14.csv",
    nrows = 20000
)

```

```python
import folium

# Centro aproximado de Nova York
mapa = folium.Map(
    location=[40.75, -73.98],
    zoom_start=11
)

# Adiciona os pontos
for _, row in df.iterrows():
    folium.CircleMarker(
        location=[row["Lat"], row["Lon"]],
        radius=1,
        fill=True
    ).add_to(mapa)

mapa.save("mapa_ny.html")
```

```python
from folium.plugins import HeatMap

mapa = folium.Map(
    location=[40.75, -73.98],
    zoom_start=11
)

HeatMap(
    df[['Lat','Lon']].values.tolist()
).add_to(mapa)

mapa
```
