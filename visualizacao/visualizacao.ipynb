# Visualização dos dados e clusters

Este notebook gera mapas interativos para apoiar a interpretação da Etapa 3. Quando `etapa3/clusters_etapa3.csv` existir, os pontos são coloridos pelo cluster final do K-Means. Caso contrário, o notebook ainda gera uma visualização básica dos dados originais.


```python
from pathlib import Path

import folium
import pandas as pd
from folium.plugins import HeatMap
from IPython.display import display

SEED = 42
AMOSTRA_MAPA = 8000

CAMINHOS_CLUSTER = [
    Path("../etapa3/clusters_etapa3.csv"),
    Path("etapa3/clusters_etapa3.csv"),
    Path("clusters_etapa3.csv"),
]

CAMINHOS_ORIGINAL = [
    Path("../ubernymaio14.csv"),
    Path("ubernymaio14.csv"),
]

SAIDA_DIR = Path(".") if Path.cwd().name == "visualizacao" else Path("visualizacao")
SAIDA_DIR.mkdir(exist_ok=True)

print("Configuração da visualização carregada. Diretório de saída:", SAIDA_DIR)
```

## 1. Carregamento

O arquivo de clusters usa as coordenadas tratadas para modelagem e, quando possível, as coordenadas originais para o mapa geográfico.


```python
cluster_path = next((caminho for caminho in CAMINHOS_CLUSTER if caminho.exists()), None)
original_path = next((caminho for caminho in CAMINHOS_ORIGINAL if caminho.exists()), None)

if cluster_path is not None:
    df = pd.read_csv(cluster_path)
    origem = f"clusters da etapa 3 ({cluster_path})"

    if {"Lat_original", "Lon_original"}.issubset(df.columns):
        lat_col, lon_col = "Lat_original", "Lon_original"
    elif original_path is not None:
        df_original = pd.read_csv(original_path, nrows=20000).drop_duplicates().reset_index(drop=True)
        if len(df_original) == len(df):
            df["Lat_original"] = df_original["Lat"].to_numpy()
            df["Lon_original"] = df_original["Lon"].to_numpy()
            lat_col, lon_col = "Lat_original", "Lon_original"
        else:
            lat_col, lon_col = "Lat", "Lon"
            print("Aviso: usando coordenadas tratadas porque o arquivo original não alinhou em quantidade de linhas.")
    else:
        lat_col, lon_col = "Lat", "Lon"
        print("Aviso: usando coordenadas tratadas porque o arquivo original não foi encontrado.")
else:
    if original_path is None:
        raise FileNotFoundError("Nenhum arquivo de clusters ou dados originais foi encontrado.")
    df = pd.read_csv(original_path, nrows=20000).drop_duplicates().reset_index(drop=True)
    df["cluster_kmeans"] = "sem cluster"
    lat_col, lon_col = "Lat", "Lon"
    origem = f"dados originais sem clusters ({original_path})"

print("Origem dos dados:", origem)
print("Dimensão:", df.shape)
print("Colunas geográficas:", lat_col, lon_col)
display(df.head())
```

## 2. Mapa colorido por cluster

Para evitar HTML excessivamente pesado, o mapa usa uma amostra reprodutível de até 8.000 pontos.


```python
df_mapa = df.sample(n=min(AMOSTRA_MAPA, len(df)), random_state=SEED).copy()
centro = [df_mapa[lat_col].median(), df_mapa[lon_col].median()]

mapa = folium.Map(location=centro, zoom_start=11, tiles="cartodbpositron")

cores = [
    "#1f77b4",
    "#d62728",
    "#2ca02c",
    "#ff7f0e",
    "#9467bd",
    "#8c564b",
    "#17becf",
    "#7f7f7f",
]

clusters_ordenados = sorted(df_mapa["cluster_kmeans"].dropna().unique(), key=lambda valor: str(valor))
cor_por_cluster = {cluster: cores[indice % len(cores)] for indice, cluster in enumerate(clusters_ordenados)}

for _, row in df_mapa.iterrows():
    cluster = row["cluster_kmeans"]
    popup = f"Cluster: {cluster}"
    if "Date/Time" in row.index:
        popup += f"<br>Data/Hora: {row['Date/Time']}"

    folium.CircleMarker(
        location=[row[lat_col], row[lon_col]],
        radius=2,
        color=cor_por_cluster.get(cluster, "#444444"),
        fill=True,
        fill_color=cor_por_cluster.get(cluster, "#444444"),
        fill_opacity=0.65,
        weight=1,
        popup=popup,
    ).add_to(mapa)

itens_legenda = "".join(
    f"<div><span style='background:{cor_por_cluster[cluster]};width:12px;height:12px;display:inline-block;margin-right:6px;'></span>Cluster {cluster}</div>"
    for cluster in clusters_ordenados
)
legenda = f"""
<div style="position: fixed; bottom: 28px; left: 28px; z-index: 9999; background: white; padding: 10px 12px; border: 1px solid #999; font-size: 13px;">
  <strong>Clusters</strong>
  {itens_legenda}
</div>
"""
mapa.get_root().html.add_child(folium.Element(legenda))

mapa_path = SAIDA_DIR / "mapa_ny.html"
mapa_clusters_path = SAIDA_DIR / "mapa_clusters_ny.html"
mapa.save(mapa_path)
mapa.save(mapa_clusters_path)

print("Mapa salvo em:", mapa_path)
print("Cópia com nome explícito salva em:", mapa_clusters_path)
mapa
```

## 3. Heatmap de densidade

O heatmap mostra a concentração espacial dos registros independentemente do cluster.


```python
mapa_heat = folium.Map(location=centro, zoom_start=11, tiles="cartodbpositron")
HeatMap(df_mapa[[lat_col, lon_col]].values.tolist(), radius=12, blur=16).add_to(mapa_heat)

heatmap_path = SAIDA_DIR / "heatmap_ny.html"
mapa_heat.save(heatmap_path)

print("Heatmap salvo em:", heatmap_path)
mapa_heat
```
